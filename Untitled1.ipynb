```python
from pathlib import Path
import preprocessing as pp
from master_pipeline import FAMILIES, build_master_table

data_dir = Path("data")

FAMILIES["cylinder_recessed"] = pp.build_family_entry(
    uptake_txt=(data_dir / "cylinder-recessed-uptake.txt").read_text(),
    volume_csv_txt=(data_dir / "cylinder-recessed-volume.csv").read_text(),
    area_csv_txt=(data_dir / "cylinder-recessed-area.csv").read_text(),
    volume_type="removed",   # <-- new: recesses REMOVE material
)

FAMILIES["square_recessed"] = pp.build_family_entry(
    uptake_txt=(data_dir / "square-recessed-uptake.txt").read_text(),
    volume_csv_txt=(data_dir / "square-recessed-volume.csv").read_text(),
    area_csv_txt=(data_dir / "square-recessed-area.csv").read_text(),
    volume_type="removed",
)

FAMILIES["cone_recessed"] = pp.build_family_entry(
    uptake_txt=(data_dir / "cone-recessed-uptake.txt").read_text(),
    volume_csv_txt=(data_dir / "cone-recessed-volume.csv").read_text(),
    area_csv_txt=(data_dir / "cone-recessed-area.csv").read_text(),
    volume_type="removed",
)

FAMILIES["square_pillar"] = pp.build_family_entry(
    uptake_txt=(data_dir / "square-pillars-uptake.txt").read_text(),
    volume_csv_txt=(data_dir / "square-pillars-volume.csv").read_text(),
    area_csv_txt=(data_dir / "square-pillars-area.csv").read_text(),
    volume_type="added",     # <-- pillars ADD material
)

FAMILIES["grooves"] = pp.build_family_entry(
    uptake_txt=(data_dir / "grooves-uptake.txt").read_text(),
    volume_csv_txt=(data_dir / "grooves-volume.csv").read_text(),
    area_csv_txt=(data_dir / "grooves-area.csv").read_text(),
    volume_type="removed",
)

master = build_master_table()
```

```python
import matplotlib.pyplot as plt
R_COLORS = plt.get_cmap("tab10").colors

import advanced_analysis as aa
```

```python
combined_series, labels = aa.build_combined_series(FAMILIES)
rank_df = aa.compute_rank_trajectory(combined_series)
```

```python
# Rank trajectory
fig, ax = plt.subplots(figsize=(12, 6))
aa.plot_rank_trajectory(rank_df, top_n=10, labels=labels, palette=R_COLORS, ax=ax)
plt.show()

# Rank volatility
volatility = aa.compute_rank_volatility(rank_df)
stab_t = aa.find_stabilization_time(volatility)
fig, ax = plt.subplots(figsize=(10, 3))
aa.plot_rank_volatility(volatility, stabilization_time=stab_t, ax=ax)
plt.show()

# Time-to-fraction
t2f = aa.build_time_to_fraction_table(FAMILIES)
fig, ax = plt.subplots(figsize=(8, 5))
aa.plot_time_to_fraction_race(t2f, top_n_per_fraction=5, ax=ax)
plt.show()

# Sensitivity — CAPACITY version (what you already ran)
sens_capacity = aa.build_sensitivity_table(master, metric="n_near_eq")
fig, ax = plt.subplots(figsize=(8, 5))
aa.plot_sensitivity_bars(sens_capacity, ax=ax)
plt.show()

# Sensitivity — KINETIC version (new, tests the actual "depth is dominant" claim)
master_k = aa.merge_time_to_fraction_into_master(master, t2f)
sens_kinetic = aa.build_sensitivity_table(master_k, metric="t50")
fig, ax = plt.subplots(figsize=(8, 5))
aa.plot_sensitivity_bars(sens_kinetic, ax=ax)
plt.show()

# Robustness of optimum
robustness = aa.compute_optimum_robustness(master, metric="n_near_eq", top_n=5)
fig, ax = plt.subplots(figsize=(7, 4))
aa.plot_optimum_robustness(robustness, ax=ax)
plt.show()

# Dominance map
dominance = aa.compute_dominance_scores(rank_df)
fig, ax = plt.subplots(figsize=(8, 6))
aa.plot_dominance_map(dominance, labels=labels, top_n=15, ax=ax)
plt.show()

# Penetration depth (pick one family)
fig, ax = plt.subplots(figsize=(8, 5))
aa.plot_penetration_vs_depth(FAMILIES["cone_recessed"]["series"], family_name="cone_recessed", ax=ax)
plt.show()

# Growth rate (fixed version — clips the t~1s spike, log y-axis)
fig, ax = plt.subplots(figsize=(8, 5))
aa.plot_growth_rate_overlay(FAMILIES["cone_recessed"]["series"], family_name="cone_recessed", palette=R_COLORS, ax=ax)
plt.show()

# Landscape evolution
fig, axes = aa.plot_landscape_evolution(
    master, "cone_recessed",
    [("n_t2000", "Early: t=2,000s"), ("n_t15000", "Mid: t=15,000s"), ("n_near_eq", "Late: near-eq")]
)
plt.show()

# Volume-normalized (fixed version — circle=removed, triangle=added)
fig, ax = plt.subplots(figsize=(7, 5))
aa.plot_volume_normalized_comparison(master, ax=ax)
plt.show()
```
